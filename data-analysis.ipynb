```python
#Function to convert the "annual_tourists" values to numeric format (ex. 12 million to 12,000,000):
def convert_tourists(value):
    value = str(value).strip()

    # Case 1: range in millions (e.g. 35-40 million)
    if "million" in value and "-" in value:
        low, high = value.replace(" million","").split("-")
        return int((float(low) + float(high)) / 2 * 1_000_000)

    # Case 2: single value in millions (e.g. 14 million)
    elif "million" in value:
        return int(float(value.replace(" million","")) * 1_000_000)

    # Case 3: already a number (thousands or hundreds)
    else:
        return int(float(value))

tourism_filtered["annual_tourists"] = tourism_filtered["annual_tourists"].apply(convert_tourists)
```

```python
#Check the unique values in "price_range" column:
restaurants_df["price_range"].unique()
```

```python
#Create a dictionary to replace the price_rnge with exploatable data
price_dict = {'$$ - $$$':'2.5', '$$$$': '4', '$':'1'}

#Replace the price_range with the values from dictionary
restaurants_df['price_range'] = restaurants_df['price_range'].replace(price_dict)
```

```python
#Convert the columns to numeric to be able to group and calculate the mean:
for col in ['rating', 'number_of_reviews', 'price_range']:
    restaurants_df[col] = pd.to_numeric(restaurants_df[col], errors='coerce')
```

```python
#Check the null or missing values:
restaurants_df.isna().sum()
```

```python
#Check the datatypes of each column:
restaurants_df.dtypes
```

```python
#Functions to fill the missing data for numeric and categorical columns:
numerical_cols = ["ranking", "rating", "price_range", "number_of_reviews"]
categorical_cols = ["city"]

def fill_missing_values(restaurants_df):

    existing_numerical = [col for col in numerical_cols if col in restaurants_df.columns]
    existing_categorical = [col for col in categorical_cols if col in restaurants_df.columns]

    if existing_numerical:
        restaurants_df.loc[:, existing_numerical] = restaurants_df[existing_numerical].fillna(
            restaurants_df[existing_numerical].mean().round(2)
        )

    if existing_categorical:
        restaurants_df.loc[:, existing_categorical] = restaurants_df[existing_categorical].fillna("Unknown")

    return restaurants_df

restaurants_df = fill_missing_values(restaurants_df)
```

```python
#Add a new column "city_id":
restaurants_df["city_id"] = restaurants_df["city"].map(city_dict)
```

```python
#Reset the index and store it as a new column called 'name_id'
restaurants_df = restaurants_df.reset_index().rename(columns={'index': 'name_id'})

#Select only the columns that will be used for the analysis:
restaurants_df = restaurants_df[["name_id", "name", "city_id", "city", "ranking", "rating", "price_range", "number_of_reviews"]]
restaurants_df
```

#### Aggregation od data and analysis in Python:

```python
#Aggregate the data for "rating", "number_of_reviews", "price_range" and calculate the average:
restaurants_summary = restaurants_df.groupby('city_id').agg({
    'rating': 'mean',
    'number_of_reviews': 'sum',
    'price_range': 'mean'
}).reset_index()
```

```python
#Round numeric columns:
restaurants_summary[['rating', 'price_range']] = restaurants_summary[['rating', 'price_range']].round(2)
```

```python
#Merge 2 datasets in one table
new_df = pd.merge(tourism_filtered, restaurants_summary, on="city_id", how="inner")
print(new_df.shape)
new_df.head(30)
```
