```python
import pandas as pd
from sqlalchemy import create_engine

# 1. Load your cleaned CSV
# Using 'utf-8-sig' specifically ignores that '0xef' BOM error we saw earlier
df = pd.read_csv("~/Downloads/city_data.csv", encoding='utf-8-sig')

# 2. Setup the Connection String
# Format: mysql+pymysql://[USER]:[PASSWORD]@[HOST]/[DATABASE_NAME]
USER = 'helenakm'
PASSWORD = 'SriLanka-2026!' # <--- Change this!
HOST = 'localhost'
DATABASE = 'tourism_project'    # <--- Change this!

# We use the variables USER, PASSWORD, HOST, and DATABASE that you defined above
engine = create_engine(f'mysql+pymysql://{USER}:{PASSWORD}@{HOST}/{DATABASE}')

# 3. Push the data to your existing table
try:
    # 'append' adds the data to your existing table structure
    df.to_sql('city_data', con=engine, if_exists='append', index=False)
    print("🚀 Success! Your data is now in MySQL.")
except Exception as e:
    print(f"❌ Error: {e}")
```

```python
# 1. Load your dataset
df = pd.read_csv("~/Downloads/restaurants.csv", encoding='utf-8-sig')

# 2. DROP the columns that don't exist in your SQL table
# We remove 'name_id' and 'city' so the DataFrame matches the SQL schema
df = df.drop(columns=['name_id', 'city'])

# 3. Push to SQL
try:
    # Now that the columns match, it should slide right in
    df.to_sql('restaurants', con=engine, if_exists='append', index=False)
    print("🚀 Success! 125,527 restaurants are traveling to MySQL.")
except Exception as e:
    print(f"❌ Error: {e}")
```

```python
# 1. Load your data
df = pd.read_csv("~/Downloads/cities.csv", encoding='utf-8-sig')

# 2. RENAME 'city' to 'name' to match your SQL table definition
# Your SQL table has (city_id, name), but your CSV has (city_id, city)
df = df.rename(columns={'city': 'name'})

# 3. Push to SQL
try:
    df.to_sql('cities', con=engine, if_exists='append', index=False)
    print("🚀 Success! The 'cities' table is now populated.")
except Exception as e:
    print(f"❌ Error: {e}")
```
